```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data3 = pd.read_csv('BT3.310092301_scoring_credit_linear.csv', delimiter=",") 
data3.head()
```

X là df chứa dữ liệu của thuộc tính Income, y là series chứa dữ liệu của thuộc tính Credit_Score

```python
X = data3[['Income']]  # Biến độc lập, cần đặt trong dấu [[]] để giữ dạng DataFrame
y = data3['Credit_Score']  # Biến phụ thuộc

```

```python
print(X)
```

```python
print(y)
```

```python


# Khởi tạo mô hình
model = LinearRegression()

# Huấn luyện mô hình
model.fit(X, y)

```

```python
# Dự đoán giá trị Credit_Score dựa trên Income
y_pred = model.predict(X)
```

```python
# Hệ số hồi quy (slope) và intercept
slope = model.coef_[0]
intercept = model.intercept_

# Tính R-square
r_square = model.score(X, y)

# In kết quả
print(f"Phương trình hồi quy: Credit_Score = {slope:.4f} * Income + {intercept:.4f}")
print(f"R-square: {r_square:.4f}")
```

```python
# Vẽ biểu đồ
plt.scatter(X, y, color='blue', label="Thực tế")  # Dữ liệu thực tế
plt.plot(X, y_pred, color='red', label="Hồi quy tuyến tính")  # Đường hồi quy
plt.xlabel("Income")
plt.ylabel("Credit_Score")
plt.title("Hồi quy tuyến tính giữa Income và Credit_Score")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
# Tính MSE và RMSE
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

# In kết quả
print(f"MSE (Mean Squared Error): {mse:.4f}")
print(f"RMSE (Root Mean Squared Error): {rmse:.4f}")
```
