```python
import pandas as pd 
from scipy import stats
```

```python
data5 = pd.read_csv('BT5.Nonlinear_data.csv', delimiter=",") 
data5.head()
```

Sử dụng thư viện scikit-learn để xây dựng mô hình hồi quy tuyến tính. Xác định giá trị y_pred từ mô hình và trực quan hóa dữ liệu , mô hình tìm được

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
x_linear = data5[['x']]
y_linear = data5['y']
```

```python

# Khởi tạo mô hình
linear_model = LinearRegression()

# Huấn luyện mô hình
linear_model.fit(x_linear, y_linear)
```

```python
y_linear_pred = linear_model.predict(x_linear)
```

```python

slope = linear_model.coef_[0]
intercept = linear_model.intercept_

r_square = linear_model.score(x_linear, y_linear)

print(f"Hệ số chặn:{intercept:.4f}")
print(f"Hệ số hồi quy slope:{slope:.4f}")
print(f"Phương trình hồi quy: y = {slope:.4f} * x + {intercept:.4f}")
print(f"R-square: {r_square:.4f}")
```

```python
# Vẽ biểu đồ
plt.scatter(x_linear, y_linear, color='blue', label="Thực tế")  # Dữ liệu thực tế
plt.plot(x_linear, y_linear_pred, color='red', label="Hồi quy tuyến tính")  # Đường hồi quy
plt.xlabel("x")
plt.ylabel("y")
plt.title("Hồi quy tuyến tính giữa x và y")
plt.legend()
plt.show()
```

```python

mse = mean_squared_error(y_linear, y_linear_pred)

rmse = np.sqrt(mse)

print(f"Hệ số chặn:{intercept:.4f}")
print(f"Hệ số hồi quy slope:{slope:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
```

Sử dụng scipy để xây dựng mô hình hồi quy phi tuyến tính (Polinominal). Trực quan hóa mô hình tìm được từ y_pred. Hàm phi tuyến: y = ax2 + bx + c.
Sử dụng curve_fit. Đầu vào x, y là array

```python
from scipy.optimize import curve_fit
```

```python

x_nonlinear = data5['x'].values  
y_nonlinear = data5['y'].values  
```

```python
def quadratic_function(x, a, b, c):
    return a * x**2 + b * x + c
```

```python
params, covariance = curve_fit(quadratic_function, x_nonlinear, y_nonlinear)

a, b, c = params
print(f"Hệ số tìm được: a = {a:.4f}, b = {b:.4f}, c = {c:.4f}")
```

```python
y_nonlinear_pred = quadratic_function(x_nonlinear, a, b, c)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(x_nonlinear, y_nonlinear, color='blue', label='Dữ liệu thực tế')

plt.plot(x_nonlinear, y_nonlinear_pred, color='red', linewidth=2, label='Mô hình phi tuyến')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Hồi quy phi tuyến (Polynomial Regression)")
plt.legend()
plt.show()
```

```python

mse = mean_squared_error(y_nonlinear, y_nonlinear_pred)

rmse = np.sqrt(mse)

print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
```
