```python
import pandas as pd 
from scipy import stats
```

```python
data = pd.read_csv('BT1.225092302_energy_drink_survey.csv', delimiter=";") 
data.head()
```

Tạo bảng chéo và in ra kết quả thống kê theo 2 thuộc tính Gender, Brand Preference. Sử dụng crosstab

```python
print(data.columns)
```

```python

crosstab_result = pd.crosstab(data['Gender'], data['Brand Preference'], margins=True)
print(crosstab_result)
```

Kiểm định Chi-square. Sử dụng chi2_contigenvy()

Giả thuyết cần kiểm tra: 
H0: Không có mối quan hệ giữa Gender và Brand Preference, tức là việc một người chọn thương hiệu nào là hoàn toàn ngẫu nhiên, không phụ thuộc vào giới tính.
H1: Có mối quan hệ giữa Gender và Brand Preference, nghĩa là giới tính có ảnh hưởng đến sở thích thương hiệu.


```python
from scipy.stats import chi2_contingency
```

```python
from scipy.stats import chi2

alpha = 0.05  # Mức ý nghĩa
df = 6  # Bậc tự do

# Tính giá trị tới hạn của Chi-square
chi_critical = chi2.ppf(1 - alpha, df)

print(f"Giá trị tới hạn của X^2(0.05,6): {chi_critical:.4f}")
```

```python
chi2, p, dof, expected = chi2_contingency(crosstab_result)

print(f"Chi-square: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies Table:")
print(pd.DataFrame(expected, index=crosstab_result.index, columns=crosstab_result.columns))
```

Kết luận: X2(Chi-square) < X2(alpha,df) => Chấp nhận H0. Giới tính không ảnh hưởng đên thương hiệu, một người chọn thương hiệu nào là hoàn toàn ngẫu nhiên, không phụ thuộc vào giới tính.
